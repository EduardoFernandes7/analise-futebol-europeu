# Análise exploratória: as 5 grandes ligas europeias

Este notebook explora dados de partidas e classificação da Premier League, La Liga, Bundesliga, Serie A e Ligue 1 (temporada 2025/26), buscados via [football-data.org](https://www.football-data.org/).

Perguntas que queremos responder:

1. Jogar em casa realmente é uma vantagem? Isso varia entre ligas?
2. Qual liga é mais ofensiva (mais gols por partida)?
3. Quão disputado foi o título em cada liga?
4. Quais times tiveram desempenho melhor ou pior do que o saldo de gols sugeriria?

```python
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent))

import plotly.express as px

from analysis.metrics import (
    finished_matches,
    goals_summary_table,
    home_advantage_table,
    load_data,
    overperformance_table,
    title_race_gap_table,
)

PROCESSED_DIR = Path.cwd().parent / "data" / "processed"
matches, standings = load_data(PROCESSED_DIR)
played = finished_matches(matches)

print(f"{len(played)} partidas concluídas em {matches['league_name'].nunique()} ligas")
played.head()
```

## 1. Vantagem de jogar em casa

Para cada liga, olhamos a proporção de vitórias do mandante, empates e vitórias do visitante entre as partidas já disputadas.

```python
home_adv = home_advantage_table(matches)
home_adv
```

```python
fig = px.bar(
    home_adv,
    x="league_name",
    y=["vitoria_mandante_pct", "empate_pct", "vitoria_visitante_pct"],
    barmode="stack",
    title="Resultado das partidas por mando de campo",
    labels={"league_name": "Liga", "value": "% das partidas", "variable": "Resultado"},
)
fig.show()
```

## 2. Qual liga é mais ofensiva?

Média de gols por partida (soma dos dois times) em cada liga.

```python
goals = goals_summary_table(matches)
goals
```

```python
fig = px.bar(
    goals,
    x="league_name",
    y="media_gols_por_partida",
    title="Média de gols por partida",
    labels={"league_name": "Liga", "media_gols_por_partida": "Gols/partida"},
)
fig.show()
```

## 3. Quão disputado foi o título?

Diferença de pontos entre o líder e o vice-líder ao final da temporada — quanto menor, mais disputada foi a corrida pelo título.

```python
title_race = title_race_gap_table(standings)
title_race
```

## 4. Sobre-performance e sub-performance

Comparamos a posição final real de cada time com a posição que o saldo de gols sugeriria ("posição esperada"). Times com diferença positiva terminaram *acima* do que o saldo de gols indicaria — geralmente sinal de eficiência ofensiva/defensiva nos jogos decisivos.

```python
overperformance = overperformance_table(standings)
print("Times que mais superaram o esperado:")
display(overperformance.head(5))

print("\nTimes que mais decepcionaram frente ao esperado:")
display(overperformance.tail(5))
```
